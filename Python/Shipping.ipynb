This notebook provides plots on shipping in terms of weight and volume

```python
# imports
import plotly.express as px
import pandas as pd
import ipywidgets as widgets
```

```python
shipping = pd.read_csv("../CSVs/_select_date_trunc_month_o_order_purchase_timestamp_timestamp_as_202609231755.csv")
```

```python
shipping_total = shipping.groupby("mon")[["product_weight_kg", "volume"]].sum().reset_index()

fig1 = px.line(shipping_total, x = "mon", y = "product_weight_kg", 
               title = "Weight Shipped (kg)", markers=True,
               labels={"mon": "Month", "product_weight_kg": "Weight Shipped (kg)"})
fig2 = px.line(shipping_total, x = "mon", y = "volume", 
               title = "Shipping Volume (cm\u00b3)", markers=True,
               labels={"mon": "Month", "volume": "Shipping Volume (cm\u00b3)"})

fig1.update_layout(font_family="Roboto Mono", template="plotly_white")
fig2.update_layout(font_family="Roboto Mono", template="plotly_white")

fig1.update_traces(line_color="black")
fig2.update_traces(line_color="black")

fig1.show()
fig2.show()
```

```python
fig1 = px.line(shipping, x = "mon", y = "product_weight_kg", color="seller_state",
                title = "Weight Shipped (kg)",
                labels={"mon": "Month", "product_weight_kg": "Weight Shipped (kg)", "seller_state": "Seller State"},
                color_discrete_sequence=px.colors.qualitative.Dark2)
fig2 = px.line(shipping, x = "mon", y = "volume", color="seller_state",
                title = "Weight Shipped (kg)",
                labels={"mon": "Month", "volume": "Shipping Volume (cm\u00b3)", "seller_state": "Seller State"},
                color_discrete_sequence=px.colors.qualitative.Dark2)


fig1.update_layout(font_family="Roboto Mono", template="plotly_white")
fig2.update_layout(font_family="Roboto Mono", template="plotly_white")

fig1.show()
fig2.show()
```

```python
states = sorted(shipping["seller_state"].unique())

picker1 = widgets.SelectMultiple(
    options=states, value=tuple(states[:3]), rows=10, description="States (Shift + Click to select/deselect)",
    style={"description_width": "initial"}
)

def show(selected):
    fig1 = px.line(shipping[shipping["seller_state"].isin(selected)], x = "mon", y = "product_weight_kg", color="seller_state",
                title = "Weight Shipped (kg)",
                labels={"mon": "Month", "product_weight_kg": "Weight Shipped (kg)", "seller_state": "Seller State"},
                color_discrete_sequence=px.colors.qualitative.Dark2)
    fig1.update_layout(font_family="Roboto Mono", template="plotly_white")
    fig1.show()

out1 = widgets.interactive_output(show, {"selected": picker1})
display(widgets.VBox([out1, picker1]))

picker2 = widgets.SelectMultiple(
    options=states, value=tuple(states[:3]), rows=10, description="States (Shift + Click to select/deselect)",
    style={"description_width": "initial"}
)

def show(selected):
    fig2 = px.line(shipping[shipping["seller_state"].isin(selected)], x = "mon", y = "volume", color="seller_state",
                title = "Weight Shipped (kg)",
                labels={"mon": "Month", "volume": "Shipping Volume (cm\u00b3)", "seller_state": "Seller State"},
                color_discrete_sequence=px.colors.qualitative.Dark2)
    fig2.update_layout(font_family="Roboto Mono", template="plotly_white")
    fig2.show()

out2 = widgets.interactive_output(show, {"selected": picker2})
display(widgets.VBox([out2, picker2]))
```
