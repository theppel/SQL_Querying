# Retention

This workbook plots retention statistics depending on how many days it took for the package to arrive

```python
# imports
import pandas as pd
import plotly.express as px
```

```python
# import roboto mono font
from IPython.display import HTML
HTML('<link href="https://fonts.googleapis.com/css2?family=Roboto+Mono&display=swap" rel="stylesheet">')
```

```python
# read in data
retention = pd.read_csv("../CSVs/_select_delivery_time_count_as_num_orders_avg_has_later_order_in_202609272004.csv")
```

### Distribution of delivery time

```python
# initialise figure
fig = px.bar(retention, x="delivery_time", y="num_orders",
             title="Distribution of Delivery Time",
             labels={"delivery_time": "Delivery Time (days)", "num_orders": "No. of Orders"})

# change font and template
fig.update_layout(font_family="Roboto Mono", template="plotly_white")

# change bar colour
fig.update_traces(marker_color="grey")

# show figure
fig.show()
```

### Retention by delivery time

```python
# initialise figure
fig = px.line(retention, x="delivery_time", y="pct_with_later_order",
             title="Distribution of Delivery Time",
             labels={"delivery_time": "Delivery Time (days)", "pct_with_later_order": "Retention (%)"})

# change font and template
fig.update_layout(font_family="Roboto Mono", template="plotly_white")

# change line colour
fig.update_traces(line_color="black")

# show figure
fig.show()
```
