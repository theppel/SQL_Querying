# Profit, Revenue, and Growth

This notebook plots Profit, Revenue, and the Quarter-on-Quarter growth of revenue and profit.

```python
# imports
import pandas as pd
import plotly.express as px
```

```python
# import roboto mono
from IPython.display import HTML
HTML('<link href="https://fonts.googleapis.com/css2?family=Roboto+Mono&display=swap" rel="stylesheet">')
```

```python
# read in data
growth = pd.read_csv("../CSVs/quarterly_202609272203.csv").rename(columns={"revenue": "Revenue", "profit": "Profit"})
```

### Quarterly Revenue and Profit

```python
# initialise figure
fig = px.line(growth, x="quarter", y=["Revenue", "Profit"], title="Quarterly Revenue and Profit",
              labels={"quarter": "Quarter", "value": "Amount ($)", "variable": ""},
              color_discrete_sequence=px.colors.qualitative.Dark2)

# make hover template
fig.update_traces(
    hovertemplate="%{fullData.name}<br>Quarter=%{x|%b %Y}<br>Amount=$%{y:,.0f}<extra></extra>"
)

# change font and template
fig.update_layout(font_family="Roboto Mono", template="plotly_white")

# show figure
fig.show()
```

### Quarter-on-Quarter Revenue and Profit Growth

Starting from once it has settled

```python
# create new data frame and rename
growth_plot = growth[["quarter", "rev_growth", "prof_growth"]].rename(columns={"rev_growth": "Revenue", "prof_growth": "Profit"})

# filter so that the initial changes are not included
growth_plot = growth_plot[growth_plot["Revenue"] < 1000]

# initialise figure
fig = px.line(growth_plot, x="quarter", y=["Revenue", "Profit"], title="Quarterly Revenue and Profit Growth",
              labels={"quarter": "Quarter", "value": "Growth (%)", "variable": ""},
              color_discrete_sequence=px.colors.qualitative.Dark2)

# make hover template
fig.update_traces(
    hovertemplate="%{fullData.name}<br>Quarter=%{x|%b %Y}<br>Amount=%{y:,.0f}%<extra></extra>"
)

# change font and template
fig.update_layout(font_family="Roboto Mono", template="plotly_white")

# add horizontal line at 0%
fig.add_hline(y=0, line_dash="dash", line_color="black")

# show figure
fig.show()
```
